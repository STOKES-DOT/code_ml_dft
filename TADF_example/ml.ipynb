```python
import joblib
import pandas as pd
import numpy as np
import os
import re
```

```python
df = pd.read_csv("/Users/jiaoyuan/Documents/GitHub/code_ml_dft/TADF_example/descriptors.csv")
```

```python
def convert_numbers_in_string(s):
    # 使用正则表达式找到字符串中的所有数字
    numbers = re.findall(r'\d+\.\d+|\d+', s)
    for number in numbers:
        # 将找到的数字替换为浮点数
        s = s.replace(number, str(float(number)))
    return s

for column in df.columns:
    # 将当前列转换为列表，以便修改
    column_data = df[column].tolist()
    # 遍历当前列中的每个值
    for i, value in enumerate(column_data):
        # 检查值是否为字符串
        if isinstance(value, str):
            # 尝试将字符串中的数字转换为浮点数
            column_data[i] = convert_numbers_in_string(value)
    # 将修改后的列数据更新回DataFrame中
    df[column] = column_data

# 输出修改后的DataFrame
print(df)

for col in df.columns:
    if df[col].dtype == 'object':  # 如果列是字符串类型
        try:
            df[col] = pd.to_numeric(df[col], errors='coerce')
        except ValueError:
            print(f"无法将列 '{col}' 转换为数值类型。")

# 删除包含缺失值的行（如果在转换过程中产生了NaN）
df.dropna(inplace=True)
```

```python
scalar_wb97xd = joblib.load("/Users/jiaoyuan/Documents/GitHub/code_ml_dft/SGM_wB97XD/Stacking_model/scaler_wb97xd.pkl")
stacking_wb97xd = joblib.load("/Users/jiaoyuan/Documents/GitHub/code_ml_dft/SGM_wB97XD/Stacking_model/final_stacking_model_wb97xd.pkl")

# 使用预处理的DataFrame进行预测
predictions_wb97xd = stacking_wb97xd.predict(scalar_wb97xd.transform(df))
df_1 = df.copy()
# 将预测结果添加到DataFrame中和
df_1['predictions_wb97xd'] = predictions_wb97xd
# 保存结果到CSV文件
output_file = "/Users/jiaoyuan/Documents/GitHub/code_ml_dft/TADF_example/predictions_wb97xd.csv"
df_1.to_csv(output_file, index=False)
print(f"预测结果已保存到 {output_file}")
```

```python
scalar_LC = joblib.load("/Users/jiaoyuan/Documents/GitHub/code_ml_dft/SGM_LC-wPBE/Stacking_model/scaler_LC.pkl")
stacking_lc_wpbe = joblib.load("/Users/jiaoyuan/Documents/GitHub/code_ml_dft/SGM_LC-wPBE/Stacking_model/final_stacking_model_LC.pkl")

# 使用预处理的DataFrame进行预测
predictions_lc_wpbe = stacking_lc_wpbe.predict(scalar_LC.transform(df))
# 将预测结果添加到DataFrame中
df_2 = df.copy()
df_2['predictions_LC'] = predictions_lc_wpbe
# 保存结果到CSV文件
output_file_lc = "/Users/jiaoyuan/Documents/GitHub/code_ml_dft/TADF_example/predictions_LC.csv"
df_2.to_csv(output_file_lc, index=False)
print(f"预测结果已保存到 {output_file_lc}")
```

```python
import lime
import lime.lime_tabular
explainer_wb97xd = lime.lime_tabular.LimeTabularExplainer(
    training_data=np.array(scalar_wb97xd.transform(df)),
    feature_names=df.columns.tolist(),
    class_names=['predictions_wb97xd'],
    mode='regression'
)
explainer_lc = lime.lime_tabular.LimeTabularExplainer(
    training_data=np.array(scalar_LC.transform(df)),
    feature_names=df.columns.tolist(),
    class_names=['predictions_LC'],
    mode='regression'
)
def explain_instance(instance_index, model, explainer, df, scalar):
    instance = df.iloc[instance_index]
    instance_scaled = scalar.transform([instance])
    explanation = explainer.explain_instance(
        data_row=instance_scaled[0],
        predict_fn=model.predict,
        num_features=10
    )
    return explanation

# 选择要解释的实例索引
instance_index = 9  # 你可以选择任何你想要解释的实例索引
explanation_wb97xd = explain_instance(instance_index, stacking_wb97xd, explainer_wb97xd, df, scalar_wb97xd)

explanation_lc = explain_instance(instance_index, stacking_lc_wpbe, explainer_lc, df, scalar_LC)

# 打印解释结果
print("LIME解释结果（wB97XD）：")
for feature, weight in explanation_wb97xd.as_list():
    print(f"{feature}: {weight}")
print("\nLIME解释结果（LC-wPBE）：")
for feature, weight in explanation_lc.as_list():
    print(f"{feature}: {weight}")
    
for instance_index in range(len(df)):
    explanation_wb97xd = explain_instance(instance_index, stacking_wb97xd, explainer_wb97xd, df, scalar_wb97xd)
    explanation_lc = explain_instance(instance_index, stacking_lc_wpbe, explainer_lc, df, scalar_LC)
    
    # 打印解释结果
    print(f"\nLIME解释结果（wB97XD） for instance {instance_index}:")
    for feature, weight in explanation_wb97xd.as_list():
        print(f"{feature}: {weight}")
    
    print(f"\nLIME解释结果（LC-wPBE） for instance {instance_index}:")
    for feature, weight in explanation_lc.as_list():
        print(f"{feature}: {weight}")
```
